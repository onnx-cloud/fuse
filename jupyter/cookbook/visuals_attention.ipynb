# Visual Cookbook: Attention Visualization 👁️

This recipe shows how to visualize attention matrices using `%attention` and `%inspect`.

```python
%load_ext src.jupyter.inspect.magics
```

## Example: multi-head attention

```python
import numpy as np

# Create a synthetic attention tensor: heads x seq x seq
att = np.random.rand(6, 16, 16).astype('float32')

# Show the attention heatmaps
%attention att
```

You can inspect specific heads or aggregate across heads using flags like `--head` or `--aggregate`.

```python
# Show only head 0
%attention att --head 0
```
