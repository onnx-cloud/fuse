# Visual Cookbook: Image Visualization 🖼️

This recipe demonstrates how to visualize image-like tensors using the `%image` magic and the `inspect` helper.

```python
# Load magics (only needed once per kernel)
%load_ext src.jupyter.magics
%load_ext src.jupyter.inspect.magics
```

## Creating an example image

```python
import numpy as np

# Create a random image HWC (height, width, channels)
img = (np.random.rand(128, 128, 3) * 255).astype('uint8')

# Use %image to render
%image img
```

You can also use the generic `%inspect` magic to auto-detect the best visualization:

```python
%inspect img
```

**Tip:** Use `--imagenet` to apply ImageNet denormalization if your tensor is normalized to mean/std.
