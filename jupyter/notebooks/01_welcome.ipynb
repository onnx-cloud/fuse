# Welcome to Fuse! 🚀

Fuse is a **cognitive compiler** for building neural networks using a simple, math-like language. 
In this tour, we'll explore how to write models without the complexity of traditional frameworks.

### Why Fuse?
- **Deterministic**: Same code, same binary ONNX output.
- **Type-Safe**: Errors are caught at compile-time, not runtime.
- **Readable**: It looks like the math it represents.

## Your First Fuse Block

In Fuse, we use the `%%fuse` magic to write code. Every Fuse file or cell starts with a version declaration `@fuse`. 

Here is a simple identity function that takes a number and returns it.

```python
%%fuse
@fuse 0.7

node main(x: f32[1]) -> f32[1] {
  x
}
```

### What happened?
1. `@fuse 0.7` sets the language version.
2. `node main(...)` defines a computational unit named `main`.
3. `x: f32[1]` means `x` is a 32-bit floating point number (like a decimal) with 1 dimension.
4. The body just returns `x`.
