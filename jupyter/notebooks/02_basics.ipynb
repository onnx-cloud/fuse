## Simple Arithmetic

You can use infix operators (`+`, `-`, `*`, `/`) or call the operators by name.

```python
%%fuse
node math_demo(a: f32[1], b: f32[1]) -> f32[1] {
  sum = a + b    # Standard addition
  diff = a - b   # Standard subtraction
  
  # We can also call ONNX operators directly
  product = Mul(a, b)
  
  # Let's return the sum
  sum
}
```

## Variable Assignment

Variables in Fuse are immutable within their scope (SSA style). You assign them once, and the compiler tracks their flow. 

Try changing the `add_one` node to multiply by two instead.

```python
%%fuse
node add_one(x: f32[1]) -> f32[1] {
  one = 1.0
  Add(x, one)
}
```
