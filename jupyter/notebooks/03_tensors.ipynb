## What is a Tensor?

A tensor is just a container for numbers. 
- `f32[1]` is a single number (Scalar/Vector size 1).
- `f32[3]` is a sequence of 3 numbers (Vector).
- `f32[2, 2]` is a grid (Matrix).

## Symbolic Dimensions (The `B` dimension)

In ML, we often process many items at once. This is called **Batching**. 
We use letters like `B` to represent a dimension that can change (variable size).

```python
%%fuse
# This node takes a batch of images (B items, 3 colors, 224x224 pixels)
node process_batch(x: f32[B, 3, 224, 224]) -> f32[B, 3, 224, 224] {
  # We can scale the whole batch by 0.5 with one operation!
  x * 0.5
}
```

### Takeaway
Tensors allow us to write high-level logic that operates on large chunks of data simultaneously. In the next tutorial, we'll see how to multiply these grids together.
