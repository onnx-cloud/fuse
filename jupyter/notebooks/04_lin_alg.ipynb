# Tutorial 04: Linear Algebra 📏

Neural networks are mostly big matrix multiplications. In Fuse, this is handled by `MatMul`.

```python
%load_ext src.jupyter.ipython
```

## Matrix Multiplication

To multiply two grids, the inner dimensions must match. 
Example: `[B, 64] * [64, 128] = [B, 128]`

```python
%%fuse
node projection(x: f32[B, 64], weights: f32[64, 128]) -> f32[B, 128] {
  # MatMul is the foundation of most layers in ML
  MatMul(x, weights)
}
```

## Transposing

Sometimes you need to flip your grid. Use `Transpose` for that.

```python
%%fuse
node flip(x: f32[10, 20]) -> f32[20, 10] {
  Transpose(x)
}
```
