# Tutorial 05: Modular Logic (Nodes) 🧱

In Fuse, we use `node` to define reusable functions. Nodes are internal to the file and help keep your code DRY (Don't Repeat Yourself).

```python
%load_ext src.jupyter.extension
```

## Defining and Calling Nodes

You can define multiple nodes and call them within each other.

```python
%%fuse
node square(x: f32[1]) -> f32[1] {
  x * x
}

node distance_squared(a: f32[1], b: f32[1]) -> f32[1] {
  diff = a - b
  square(diff)
}
```

## Recursion and Loops?

Fuse is a **Graph DSL**. This means there is no traditional recursion. Logic flows from top to bottom. For repeating patterns, we use specific operators or higher-level abstractions which we'll cover later.
