# Tutorial 06: The Graph (Exporting) 📦

While `node` is for internal logic, `graph` is for defining the **Exportable Model**. 
A `graph` becomes an entry point in the final ONNX model.

```python
%load_ext src.jupyter.ipython
```

## Graph Inputs and Outputs

A graph defines the external interface of your model.

```python
%%fuse mymodel
graph MyModel(input_ids: i32[1, 32]) -> f32[1, 10] {
  # Logic here...
  # For now, let's just return a dummy result
  Identity(Cast(input_ids, to=1)) # Cast to float
}
```

```python
# Display the exported graph using the Fuse visualization magics
%fuse.show mymodel
```

## Metadata

You can also add metadata like `@domain` or `@opset` at the top of a cell to control how the graph is compiled.

```python
%%fuse
@domain custom.logic
@opset onnx 15

graph SimpleAdd(x: f32[1]) -> f32[1] {
  Add(x, 1.0)
}
```
