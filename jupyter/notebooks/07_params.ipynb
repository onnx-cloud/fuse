# Tutorial 07: Parameters & Constants 💾

In ML, we have **Inputs** (data that changes) and **Parameters** (weights that are learned). 
Fuse distinguishes between them using `@param`, `@const`, and `@train`.

```python
%load_ext src.jupyter.ipython
```

## Constants

Use `const` for values that never change. The compiler folds these into the graph.

```python
%%fuse
const PI: f32 = 3.14159

node circular_area(r: f32[1]) -> f32[1] {
  PI * r * r
}
```

## Trainable Weights

Use `@train weight` for parameters that will be updated during training.

```python
%%fuse
@train weight W: f32[64, 32]
@train weight b: f32[32]

graph DenseLayer(x: f32[B, 64]) -> f32[B, 32] {
  Add(MatMul(x, W), b)
}
```
