# Tutorial 08: Shape Manipulation 🔄

Data often needs to be reshaped or combined. Fuse makes this explicit and type-safe.

```python
%load_ext src.jupyter.extension
```

## Reshaping

You can change the dimensions of a tensor as long as the total number of elements stays the same.

```python
%%fuse
node flatten(x: f32[B, 3, 32, 32]) -> f32[B, 3072] {
  Reshape(x, [B, 3072])
}
```

## Concatenation

Glue multiple tensors together along an axis.

```python
%%fuse
node combine(a: f32[B, 16], b: f32[B, 16]) -> f32[B, 32] {
  Concat(a, b, axis=1)
}
```
