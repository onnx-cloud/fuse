# Tutorial 09: Proofs & Safety ✅

How do you know your model is correct? Fuse has built-in testing called **Proofs**.

```python
%load_ext src.jupyter.ipython
```

## @proof

A `@proof` is a special block that runs at compile-time to verify shapes and values.

```python
%%fuse
node add_five(x: f32[1]) -> f32[1] {
  x + 5.0
}

@proof graph test_add_five() {
  input = [10.0]
  output = add_five(input)
  
  # If this assert fails, the code won't compile!
  assert output == [15.0]
}
```

### Why use proofs?
- Catch dimension mismatches before deploying.
- Document the expected behavior of your nodes.
- Ensure your logic matches the math description.
