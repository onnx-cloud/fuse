# Tutorial 10: Reusing Models 🌐

You don't always have to build from scratch. Fuse allows you to **Import** existing ONNX models and fuse them into your logic.

```python
%load_ext src.jupyter.ipython
```

## @import

You can import a model from a local file or a remote URL.

```python
%%fuse
# Import a pre-trained sentiment analysis model
@import nlp.classifier as Sentiment from "./weights/sentiment.onnx"

graph AnalyzeText(ids: i32[1, 32]) -> f32[1, 3] {
  # Use the imported model as if it were a local node!
  Sentiment(ids)
}
```
